```python
import numpy as np                                       #  (1)
import matplotlib.pyplot as plt                          #  (2)
ordinates = []                                           #  (3)
abscissas = np.arange(0.01, 2.00, 0.01)                  #  (4)
arc_ab = 0                                               #  (5)
for abscissa in abscissas:                               #  (6)
    sine_gb = np.sqrt(2*abscissa - abscissa**2)          #  (7)
    arc_ab += (0.01)/(np.sqrt(2*abscissa - abscissa**2)) #  (8)
    ordinates.append(sine_gb + arc_ab)                   #  (9)
plt.figure(figsize=(2.00,3.14))                          # (10)
plt.plot(abscissas, ordinates)                           # (11)
plt.show()                                               # (12)

# ---------------------------------------

# SETUP:
# (1) A function from the 'numpy' package is used to get a sequence
#     of numbers evenly spaced over a certain range.
# (2) The 'matplotlib' package is used to draw and display the graph.
# (3) This defines an empty list, which will hold the ordinates.
# (4) This defines the list of abscissas corresponding to the ordinates, 
#     that is, the 'x' values or lengths along the diameter AD. The list 
#     is composed of all the values from 0.01 to 1.99 in increments of 0.01. 
# (5) The symbol 'arc_ab' is a name for the running total of the arc length 
#     values being summed.

# CALCULATION:
# (6) This sets up the main loop, performaing a calculation to generate an
#     ordinate for each abscissa.
# (7) Each ordinate is the sum of GB and AB and we start by calculating GB, 
#     whose length is the square root of twice the coordinate minus its square.
# (8) We "integrate" by treating dx as 0.01 and adding the integrand to the total.
# (9) having calculated a total AB + GB for a particular x, we add it
#     to our list of ordinates

# DRAWING:
# (10) We set up the axes of the picture to use the same scale as the data,
# (11) draw a graph of the ordinates (vertically) for each abscissa,
# (12) and display it.

```
