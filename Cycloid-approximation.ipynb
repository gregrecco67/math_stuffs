```python
import numpy as np                                       # (1)
import matplotlib.pyplot as plt                          # (2)
ordinates = []                                           # (3)
abscissas = np.arange(0.01, 2.00, 0.01)                  # (4)

for abscissa in abscissas:                               # (5)
    total = np.sqrt(2*abscissa - abscissa**2)            # (6)
    for x in np.arange(0.0001, abscissa+0.0001, 0.0001): # (7)
        total += (0.0001)/(np.sqrt(2*x - x**2))          # (8)
    ordinates.append(total)                              # (9)

plt.figure(figsize=(2.00,3.14))                          # (10)
plt.plot(abscissas, ordinates)                           # (11)
plt.show()                                               # (12)

# ---------------------------------------

# (1) We will use a function from this package to get an increasing
# sequence of numbers evenly spaced over a certain range.

# (2) We use this package to draw and display the graph.

# (3) This defines an empty list, which we will fill with the values
# we calculate for each ordinate.

# (4) This defines the list of abscissas of the ordinates, that is, the 'x'
# values or lengths along the diameter AD. The list is composed of
# all the values from 0.01 to 1.99 in increments of 0.01. These are
# the two hundred ordinates we are "sampling" in order to draw the curve

# (5) This is the main loop, performaing a calculation to generate an
# ordinate for each abscissa

# (6) Each ordinate is the sum of GB and AB and we start by calculating GB, 
# whose length is the square root of twice the coordinate minus its square.
# We call this 'total' because we'll AB to it next.

# (7) We will "integrate" by treating dx as 0.0001. This line
# defines a loop that will be run as many times as needed for each
# value of the abscissa 'x' between 0.01 and 2 (as defined on line 4)

# (8) each pass through the loop, we add the value of dx / sqrt(etc.) to the 
# total. This is the operation indicated by the "sum" sign.

# (9) having calculated a total AB + GB for a particular x, we add it
# to our list of ordinates

# (10) We set up the axes of the picture to use the same scale as the data,
# (11) draw a graph of the ordinates (vertically) for each abscissa,
# (12) and display it.

```
